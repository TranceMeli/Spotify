```python
df.loc[df["key"] != "Unknown", "key"].value_counts(normalize=True).head(5) * 100
```

```python
# Top 5 Keys insgesamt
top_keys = df.loc[df["key"] != "Unknown", "key"].value_counts().head(5).index

# Major/Minor-Verteilung
key_mode_counts = pd.crosstab(
    df.loc[df["key"] != "Unknown", "key"],
    df["mode"]
)

# Nur Top 5 Keys
key_mode_counts = key_mode_counts.loc[top_keys]

# Prozentwerte innerhalb jedes Keys
key_mode_percent = (
    key_mode_counts
    .div(key_mode_counts.sum(axis=1), axis=0)
    .mul(100)
)

print(key_mode_counts)
print()
print(key_mode_percent.round(1))
```

Danceability describes how suitable a track is for dancing based on its musical characteristics, particularly rhythm, beat strength and tempo. The value ranges from 0 to 100 in this dataset, with higher values indicating greater danceability.

```python
top_10_danceability = (
    df[["track_name", "artist(s)_name", "danceability_%"]]
    .sort_values("danceability_%", ascending=False)
    .head(10)
)

top_10_danceability
```

```python
last_10_danceability = (
    df[["track_name", "artist(s)_name", "danceability_%"]]
    .sort_values("danceability_%", ascending=False)
    .tail(10)
)

last_10_danceability
```
