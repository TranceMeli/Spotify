```python
key_mode_counts = (
    df.groupby(["key", "mode"])
      .size()
      .unstack(fill_value=0)
)
```

```python
df["key"].value_counts()
```

```python
df["key"].value_counts().head(5)
```

```python
key_mode_counts = pd.crosstab(df["key"], df["mode"])

key_mode_counts.columns = ["Major", "Minor"]

top_keys = df["key"].value_counts().head(5).index

key_mode_counts = key_mode_counts.loc[top_keys]

key_mode_percent = key_mode_counts.div(
    key_mode_counts.sum(axis=1),
    axis=0
) * 100
```

```python
print(df["mode"].value_counts())
print()
print(key_mode_counts)
print()
print(key_mode_percent)
```

```python
print("Mean:", round(df["bpm"].mean(), 2), "BPM")
print("Median:", round(df["bpm"].median(), 2), "BPM")
```

### Tempo

Das durchschnittliche Tempo der Songs liegt bei **122,54 BPM**, während der Median bei **121 BPM** liegt. Die geringe Differenz von 1,54 BPM deutet darauf hin, dass extreme BPM-Werte den Mittelwert nur geringfügig beeinflussen.

Der Median beschreibt damit das typische Tempo des Datensatzes robuster gegenüber Ausreißern und liegt bei **121 BPM**.
