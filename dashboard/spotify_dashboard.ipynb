# Spotify Songs 2023 – Dashboard

Analysis of the **Top Spotify Songs 2023** dataset from [Kaggle](https://www.kaggle.com/datasets/nelgiriyewithana/top-spotify-songs-2023).

The charts and insights live in `src/spotify_analysis/charts.py`. The standalone HTML version (header, KPIs, footer) is built with `python scripts/build_dashboard.py` and saved as `dashboard/spotify_dashboard.html`.

> Streams are cumulative and not limited to 2023, so older songs have a head start. Key and mode charts only include songs with a known key.

```python
import sys

sys.path.insert(0, "../src")

from IPython.display import HTML, display

from spotify_analysis import charts
from spotify_analysis.data import load_clean
from spotify_analysis.report import CARDS

df = load_clean()
df.shape
```

```python
for value, label in charts.kpis(df):
    print(f"{value:>9}  {label}")
```

```python
texts = charts.insights(df)

for key, title, make_fig, _ in CARDS:
    display(HTML(f"<h3>{title}</h3>"))
    make_fig(df).show()
    display(HTML(f"<p>{texts[key]}</p>"))
```
